```python
roll_number = 1024170027
```

```python
digits = [int(d) for d in str(roll_number)]
L = [digit * 10 for digit in digits]
print("Original L:", L)
```

```python
L.append(50)
print("After append(50):", L)
```

```python
L.insert(2, 80)
print("After insert(2, 80):", L)
```

```python
L.pop(2)
print("After pop(2):", L)
```

```python
L.sort()
print("Ascending order:", L)
```

```python
L.sort(reverse=True)
print("Descending order:", L)
```

```python
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])
```

```python
average = sum(L) / len(L)

greater_than_average = [x for x in L if x > average]

print("Average:", average)
print("Elements greater than average:", greater_than_average)
```

```python
scores = tuple(L[:8])
print("Scores tuple:", scores)
```

```python
highest = max(scores)
highest_index = scores.index(highest)
```

```python
lowest = min(scores)
lowest_count = scores.count(lowest)
```

```python
print("Highest score:", highest)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest)
print("Number of times lowest appears:", lowest_count)
```

```python
reversed_scores = list(scores[::-1])
print("Reversed tuple as list:", reversed_scores)
```

```python
user_score = int(input("Enter a score to search: "))

if user_score in scores:
    print("First occurrence index:", scores.index(user_score))
else:
    print("Score is not present in the tuple.")
```

```python
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
```

```python
first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
import random
from collections import Counter
```

```python
random.seed(1024170027)
```

```python
random_numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(random_numbers)
```

```python
odd_numbers = [x for x in random_numbers if x % 2 != 0]

print("Odd numbers:")
print(odd_numbers)

print("Count of odd numbers:", len(odd_numbers))
```

```python
even_numbers = [x for x in random_numbers if x % 2 == 0]

print("\nEven numbers:")
print(even_numbers)

print("Count of even numbers:", len(even_numbers))
```

```python
def is_prime(n):
    if n < 2:
        return False
    
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    
    return True
prime_numbers = [x for x in random_numbers if is_prime(x)]

print("Prime numbers:")
print(prime_numbers)

print("Count of prime numbers:", len(prime_numbers))
```

```python
frequency = Counter(random_numbers)

most_common_number, most_common_count = frequency.most_common(1)[0]

print("Most frequently occurring number:", most_common_number)
print("Number of occurrences:", most_common_count)
```

```python
first_8_digits = digits[:8]

print("First 8 digits:", first_8_digits)

# Create sets
A = {digit * 7 for digit in first_8_digits}
B = {digit * 9 for digit in first_8_digits}

print("Set A:", A)
print("Set B:", B)
```

```python
union = A.union(B)

print("Union of A and B:", union)
```

```python
intersection = A.intersection(B)

print("Intersection of A and B:", intersection)
```

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)
```

```python
symmetric_diff = A.symmetric_difference(B)

print("Symmetric difference:", symmetric_diff)
```

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

```python
X = int(input("Enter a value X to remove from A: "))

A.discard(X)

print("Set A after discard:", A)
```

```python
dict_pop = my_dict.copy()

removed_value = dict_pop.pop("branch")

print("Using pop():")
print(dict_pop)
print("Returned value:", removed_value)
```
